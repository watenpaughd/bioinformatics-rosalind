```python
#Prob 3:
#given a, b (a < b) find the sum of all odd integers between a and b inclusive

def prob3(a, b):
  i = a
  sum = 0
  while(i <= b):
    if(i % 2 == 1):
      sum += i
    i += 1
  return sum
```

```python
#tests for 3
assert prob3(100, 200) == 7500, "test1 failed"
assert prob3(1, 3) == 4, "test2 failed"
```

```python
#prob 5

def prob5(s):
  word_list = s.split(" ")
  my_dict = {}
  for word in word_list:
    if word in my_dict:
      my_dict[word] += 1
    else:
      my_dict[word] = 1
  for key, value in my_dict.items():
    print(f"{key} {value}")
```

```python
#problem 6
#given a string of actg determine the number of as cs ts and gs in it.

def prob6(s):
  s = s.lower()
  As = s.count('a')
  Cs = s.count('c')
  Ts = s.count('t')
  Gs = s.count('g')
  return(f"{As} {Cs} {Gs} {Ts}")
```

```python
#problem 7
#translate dna to rna, aka take a string and turn Ts into Us!

def prob7(s):
  output = ""
  for char in s:
      if char == "T":
        output += "U"
      else:
        output += char
  return(output)
```
